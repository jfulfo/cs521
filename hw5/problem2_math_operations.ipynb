```python
from syncode import Syncode
```

```python
def run_syncode(textual_query):
    grammar = """start: operand ", " operator ", " operand

operand: NUMBER

operator: "+" | "-" | "*" | "/"

NUMBER: /-?[0-9]+\.?[0-9]*/
"""
    
    model_name = "microsoft/phi-2"
    syn_llm = Syncode(
        model=model_name,
        grammar=grammar,
        parse_output_only=True,
        max_new_tokens=50
    )
    
    prompt = """Extract the two operands and operator from the mathematical query. Output format: operand1, operator, operand2

Query: What is 15.5 plus 7.3?
Extraction: 15.5, +, 7.3

Query: What is 100.0 minus 25.5?
Extraction: 100.0, -, 25.5

Query: What is 8.0 multiplied by 6.5?
Extraction: 8.0, *, 6.5

Query: What is 50.0 divided by 2.5?
Extraction: 50.0, /, 2.5

Query: """ + textual_query + """
Extraction: """
    
    output = syn_llm.infer(prompt)[0]
    
    parts = [p.strip() for p in output.strip().split(',')]
    
    if len(parts) != 3:
        raise ValueError(f"Unexpected output format: {output}")
    
    operand1 = float(parts[0])
    operator = parts[1]
    operand2 = float(parts[2])
    
    if operator == '+':
        result = operand1 + operand2
    elif operator == '-':
        result = operand1 - operand2
    elif operator == '*':
        result = operand1 * operand2
    elif operator == '/':
        result = operand1 / operand2
    else:
        raise ValueError(f"Unknown operator: {operator}")
    
    return ([operand1, operand2], operator, result)
```

```python
query1 = "What is 327. multiplied by 11.0?"
result1 = run_syncode(query1)
print(f"Query: {query1}")
print(f"Operands: {result1[0]}, Operator: {result1[1]}, Final Answer: {result1[2]}, Expected: {327.0 * 11.0}")
```

```python
query2 = "What is 45.1 plus 23.54?"
result2 = run_syncode(query2)
print(f"Query: {query2}")
print(f"Operands: {result2[0]}, Operator: {result2[1]}, Final Answer: {result2[2]}, Expected: {45.1 + 23.54}")
```
